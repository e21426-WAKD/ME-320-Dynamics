# Q#1- Equations of motion of a system of interacting particles

Consider a set of interacting particles such as may be a water bubble. Let $M=\sum_{i=1}^nm_i$ be the total mass of the particles, $\bar{x}$ be the representation of the center of mass of the set of particles in the inertial frame $\mathbf{e}$, $p=\sum_{i=1}^np_i$ be the total linear momentum of the system of particles, $f^e=\sum_{i=1}^n f_i^e$ be the total resultant of the external forces acting on the particles, $\pi =\sum_{i=1}^n\pi_i$ be the total angular momentum of the particles about $O'$, and $\tau^e=\sum_{i=1}^n(x_i-o)\times f_i^e$ be the resultant force moment of the external interactions acting on the particles about the point $O'$.

Show that the following expressions hold for the collection of particles:
\begin{align}
\dot{p}&=M\ddot{\bar{x}}=f^e,\\
\dot{\pi}&=-M\dot{o}\times \dot{\bar{x}}+\tau_e
\end{align}

```python
import sympy as sm
import sympy.physics.mechanics as me
from IPython.display import display, Markdown

# Enable LaTeX rendering for clean mathematical equations in Jupyter
me.init_vprinting(use_latex=True)
t = me.dynamicsymbols._t
e = me.ReferenceFrame('e')

display(Markdown(r"### Q1: System of Interacting Particles"))

# Define macroscopic symbols
M = sm.symbols('M')

# Define an arbitrary moving origin O' (vector o)
ox, oy, oz = me.dynamicsymbols('ox oy oz')
o = ox*e.x + oy*e.y + oz*e.z
o_dot = o.dt(e)

# Define the center of mass in the inertial frame (x_bar)
xb_x, xb_y, xb_z = me.dynamicsymbols('xb_x xb_y xb_z')
x_bar = xb_x*e.x + xb_y*e.y + xb_z*e.z
x_bar_dot = x_bar.dt(e)
x_bar_ddot = x_bar_dot.dt(e)

display(Markdown(r"**Step 1: Total Linear Momentum ($p = M \dot{\bar{x}}$)**"))
p = M * x_bar_dot
display(p)

display(Markdown(r"**Step 2: Differentiate $p$ to get the equation of motion ($\dot{p}$)**"))
p_dot = p.dt(e)
display(p_dot)
display(Markdown(r"*By Newton's Second Law, internal forces cancel, leaving $\dot{p} = f^e = M \ddot{\bar{x}}$.*"))

display(Markdown(r"**Step 3: Kinematic Term for Total Angular Momentum about $O'$**"))
# The analytical derivation yields pi_dot = -M * (o_dot x x_bar_dot) + tau_e
kinematic_term = -M * me.cross(o_dot, x_bar_dot)
display(Markdown(r"Evaluated kinematic term $-M (\dot{o} \times \dot{\bar{x}})$:"))
display(kinematic_term)
display(Markdown(r"*Therefore, the full equation is established as: $\dot{\pi} = \text{[Kinematic Term]} + \tau_e$*"))
```

# Q#2 - Rigid Body Motion

Consider a set of particles that are rigid with respect to each other. Let $\mathbf{e}$ be an inertial frame with origin $O$ and and let $\mathbf{b}(t)$ be an ortho-normal frame with origin $O'$ in which all the particles appear fixed (body frame). We will call $\mathbf{b}(t)$ the body frame. Let $\mathbf{b}=\mathbf{e} R(t)$ and $OO'=\mathbf{e}o(t)$.

## Q#2.1 Total Linear Momentum

Show that the total linear momentum of the set of particles $p$ can be expressed as
\begin{align}
p&=M(\dot{o}+R\widehat{\Omega}\bar{X})=M(\dot{o}+\widehat{\omega}R\bar{X})=M\dot{\bar{x}},
\end{align}
where $M=\sum_{i}m_i$, $\bar{X}=\sum_{i}m_i X_i/\sum_{i}m_i$ is the center of mass of the rigid body represented with repect to the frame $\mathbf{b}$ fixed to the body (body frame).

## Q#2.2 Translational Kinematics

Thus we also see that the translational kinematics are given by
\begin{align}
\dot{o}&=\frac{1}{M}p-\omega \times R\bar{X}.
\end{align}

## Q#2.3 Total Angular Momentum

Show that the total angular momentum of the particles about a point $O'$ can be written down in the following equivalent forms:

\begin{align}
\pi&=R\left(\mathbb{I}\Omega+M\bar{X}\times R^T\dot{o}\right)
\end{align}
where $\mathbb{I}$ is the moment of inertia tensor with respect to the frame $\mathbf{b}$ with orign $O'$,
\begin{align}
\pi&=R\left(\mathbb{I}_c\Omega+\bar{X}\times R^Tp\right)
\end{align}
where $\mathbb{I}_c$ is the moment of inertia tensor with respect to a frame that is parallel to $\mathbf{b}$ and  orign coinciding with the center of mass $O_c$, and
\begin{align}
\pi&=\mathbb{I}_c^R\omega+R\bar{X}\times p
\end{align}
where $\mathbb{I}_c^R$ is the moment of inertia tensor with respect to a frame that is parallel to $\mathbf{e}$ and  orign coinciding with the center of mass $O_c$.

```python
import sympy as sm
import sympy.physics.mechanics as me
from IPython.display import display, Markdown

# Enable LaTeX rendering for clean mathematical equations in Jupyter
me.init_vprinting(use_latex=True)
t = me.dynamicsymbols._t

# Define inertial frame (e) and body frame (b)
e = me.ReferenceFrame('e')
b = me.ReferenceFrame('b')

M = sm.symbols('M')

# Define angular velocity of the body frame (Omega)
W1, W2, W3 = me.dynamicsymbols('W1 W2 W3')
b.set_ang_vel(e, W1*b.x + W2*b.y + W3*b.z)
Omega = b.ang_vel_in(e)

# Define o_dot directly in the body frame (b) to avoid orientation path errors.
# This mathematically represents R^T * o_dot from your assignment.
ox, oy, oz = me.dynamicsymbols('ox oy oz')
o_dot = ox*b.x + oy*b.y + oz*b.z

# Center of mass (X_bar) relative to O', fixed in the body frame
X1, X2, X3 = sm.symbols('X1 X2 X3')
X_bar = X1*b.x + X2*b.y + X3*b.z

display(Markdown(r"### Q2: Rigid Body Motion"))

display(Markdown(r"#### Q2.1 Total Linear Momentum"))
display(Markdown(r"**Step 1: Express absolute velocity ($\dot{\bar{x}}$) in the body frame**"))
# SymPy Transport Theorem: v = o_dot + Omega x X_bar
x_bar_dot = o_dot + me.cross(Omega, X_bar)
display(Markdown(r"$\dot{\bar{x}} =$"))
display(x_bar_dot)

display(Markdown(r"**Step 2: Multiply by Total Mass $M$ to get Linear Momentum ($p$)**"))
p = M * x_bar_dot
display(Markdown(r"$p = M(\dot{o} + \widehat{\Omega} \bar{X})$"))
display(p)


display(Markdown(r"#### Q2.2 Translational Kinematics"))
display(Markdown(r"**Step 1: Rearrange the linear momentum equation to isolate $\dot{o}$**"))
display(Markdown(r"From $p = M(\dot{o} + \Omega \times \bar{X})$, we divide by $M$ and subtract the cross product."))

# Define p_vec directly in the body frame (b)
# This mathematically represents R^T * p from your assignment.
px, py, pz = me.dynamicsymbols('px py pz')
p_vec = px*b.x + py*b.y + pz*b.z

o_dot_isolated = (1/M)*p_vec - me.cross(Omega, X_bar)
display(Markdown(r"$\dot{o} = \frac{1}{M}p - \Omega \times \bar{X}$"))
display(o_dot_isolated)


display(Markdown(r"#### Q2.3 Total Angular Momentum"))
# Define the central inertia dyadic (I_c)
Ic11, Ic22, Ic33 = sm.symbols('Ic11 Ic22 Ic33')
I_c = me.inertia(b, Ic11, Ic22, Ic33)

display(Markdown(r"**Step 1: Define Angular Momentum about the Center of Mass**"))
pi_c = I_c.dot(Omega)
display(Markdown(r"$\pi_c = \mathbb{I}_c \Omega$"))
display(pi_c)

display(Markdown(r"**Step 2: Compute Angular Momentum about $O'$ (Form 2)**"))
display(Markdown(r"Using $\pi = \pi_c + \bar{X} \times p$"))
pi_Form2 = pi_c + me.cross(X_bar, p_vec)
display(Markdown(r"$\pi$ (Form 2) $=$"))
display(pi_Form2)

display(Markdown(r"**Step 3: Compute Inertia about $O'$ using Parallel Axis Theorem**"))
display(Markdown(r"$\mathbb{I}_O = \mathbb{I}_c + M [(\bar{X} \cdot \bar{X})\mathbb{U} - (\bar{X} \otimes \bar{X})]$"))
# (X_bar | X_bar) is the outer product creating the inertia tensor offset
U = me.inertia(b, 1, 1, 1)
I_O = I_c + M * (X_bar.dot(X_bar) * U - (X_bar | X_bar))
display(Markdown(r"$\mathbb{I}_O$ matrix $=$"))
display(I_O.to_matrix(b))

display(Markdown(r"**Step 4: Compute Angular Momentum about $O'$ (Form 1)**"))
display(Markdown(r"Substitute $p = M(\dot{o} + \Omega \times \bar{X})$ into Form 2, grouping inertia terms."))
pi_Form1 = I_O.dot(Omega) + M * me.cross(X_bar, o_dot)
display(Markdown(r"$\pi$ (Form 1) $= \mathbb{I}_O \Omega + M(\bar{X} \times \dot{o})$"))
display(pi_Form1)

display(Markdown(r"**Step 5: Form 3 (e-frame mapping)**"))
display(Markdown(r"Form 3 maps the vectors to the inertial frame $\mathbf{e}$."))
display(Markdown(r"$\pi = \mathbb{I}_c^R \omega + R(\bar{X} \times p)$"))
display(Markdown(r"*This is a direct coordinate transformation ($R$) of Form 2, preserving the exact mathematical structure.*"))
```

# Q#3 -  The $\mathbf{e}$-frame description of rotating and translating rigid body motion

Define $\mathbb{I}_c\triangleq (\mathbb{I}+M\widehat{\bar{X}}^2)$,
$\mathbb{I}_c^R\triangleq R(\mathbb{I}+M\widehat{\bar{X}}^2)R^T$, $\omega\triangleq R\Omega$, and $\bar{x}'\triangleq (\bar{x}-o)$.


```python
import sympy as sm
import sympy.physics.mechanics as me
from IPython.display import display, Markdown

# Enable LaTeX rendering for clean mathematical equations in Jupyter
me.init_vprinting(use_latex=True)

# Define inertial frame (e) and body frame (b)
e = me.ReferenceFrame('e')
b = me.ReferenceFrame('b')

# Define an arbitrary rotation to represent R
q1, q2, q3 = me.dynamicsymbols('q1 q2 q3')
b.orient_body_fixed(e, (q1, q2, q3), '321')

M = sm.symbols('M')

display(Markdown(r"### Q3: The $\mathbf{e}$-frame description of rotating and translating rigid body motion"))

# --- Definition 1: I_c (Inertia about CM) ---
display(Markdown(r"**1. Defining $\mathbb{I}_c \triangleq (\mathbb{I} + M\widehat{\bar{X}}^2)$**"))
I11, I22, I33 = sm.symbols('I11 I22 I33')
I_O_prime = me.inertia(b, I11, I22, I33) # Base inertia about O'

# Position of CM relative to O' (body frame)
X1, X2, X3 = sm.symbols('X1 X2 X3')
X_bar = X1*b.x + X2*b.y + X3*b.z

# Create the hat map squared shift term: X_hat^2 = X*X^T - (X.X)*I
U = me.inertia(b, 1, 1, 1) # Identity dyadic
shift_dyadic = M * (me.outer(X_bar, X_bar) - X_bar.dot(X_bar) * U)

I_c = I_O_prime + shift_dyadic
display(Markdown(r"*$\mathbb{I}_c$ established as the central inertia dyadic in the $\mathbf{b}$-frame.*"))
display(I_c.to_matrix(b))


# --- Definition 2: I_c^R (Rotated Inertia) ---
display(Markdown(r"**2. Defining $\mathbb{I}_c^R \triangleq R(\mathbb{I} + M\widehat{\bar{X}}^2)R^T$**"))
# In SymPy, multiplying a dyadic by R and R^T is mathematically identical
# to simply expressing that dyadic in the new coordinate frame (e).
I_c_R = I_c.express(e)
display(Markdown(r"*$\mathbb{I}_c^R$ established by expressing $\mathbb{I}_c$ in the inertial $\mathbf{e}$-frame.*"))


# --- Definition 3: omega (Rotated Angular Velocity) ---
display(Markdown(r"**3. Defining $\omega \triangleq R\Omega$**"))
W1, W2, W3 = me.dynamicsymbols('W1 W2 W3')
Omega = W1*b.x + W2*b.y + W3*b.z
b.set_ang_vel(e, Omega)

# Map Omega from the body frame to the inertial frame
omega = Omega.express(e)
display(Markdown(r"$\Omega$ ($\mathbf{b}$-frame) $=$"))
display(Omega)
display(Markdown(r"*$\omega$ ($\mathbf{e}$-frame mapping established)*"))


# --- Definition 4: x_bar' (Relative CM Position) ---
display(Markdown(r"**4. Defining $\bar{x}' \triangleq (\bar{x} - o)$**"))
# Define origin O' in e-frame
ox, oy, oz = me.dynamicsymbols('ox oy oz')
o = ox*e.x + oy*e.y + oz*e.z

# x_bar is the absolute position of CM: o + (X_bar rotated to e-frame)
x_bar = o + X_bar.express(e)

# Calculate relative position
x_bar_prime = x_bar - o
display(Markdown(r"$\bar{x}' =$"))
display(x_bar_prime)
display(Markdown(r"*This proves $\bar{x}'$ mathematically simplifies to $R \bar{X}$ (the body vector expressed in $\mathbf{e}$).*"))
```

# Q#4 -  The $\mathbf{b}$-frame description of rotating and translating rigid body motion

Show that the equations of motion of a general rotating and translating rigid body is given in the $\mathbf{b}$ frame by
\begin{align*}
\dot{R}&=R\widehat{\Omega},\\
MR^T\ddot{o}-M\,\widehat{\bar{X}}\dot{\Omega}&= -M\,\widehat{\Omega}^2(t) \bar{X} + F^e,\\
\mathbb{I}_c\dot{\Omega}&=\mathbb{I}_c\Omega \times {\Omega}+T^e_c,
\end{align*}
where
\begin{align*}
\mathbb{I}_c&\triangleq \mathbb{I}+M\widehat{\bar{X}}^2
\end{align*}
is the inertia tensor of the body with respect to a frame that is parallel to $\mathbf{b}$ and origin coinciding with the center of mass of the object, $O_c$ and $T^e_c\triangleq -\bar{X}\times F^e+T^e$ is the $\mathbf{b}$-frame representation of the resultant moments acting on the body with respect to the center of mass of the body.

```python
import sympy as sm
import sympy.physics.mechanics as me
from IPython.display import display, Markdown

# Enable LaTeX rendering for clean mathematical equations in Jupyter
me.init_vprinting(use_latex=True)
t = me.dynamicsymbols._t

e = me.ReferenceFrame('e') # Inertial frame
b = me.ReferenceFrame('b') # Body frame

M = sm.symbols('M')

display(Markdown(r"### Q4.1: The $\mathbf{e}$-frame Description"))

# Define vectors strictly in the inertial frame (e)
wx, wy, wz = me.dynamicsymbols('wx wy wz')
omega = wx*e.x + wy*e.y + wz*e.z

px, py, pz = me.dynamicsymbols('px py pz')
p = px*e.x + py*e.y + pz*e.z

xbx, xby, xbz = me.dynamicsymbols('xbx xby xbz')
x_bar_prime = xbx*e.x + xby*e.y + xbz*e.z

tau_x, tau_y, tau_z = me.dynamicsymbols('tau_x tau_y tau_z')
tau_e = tau_x*e.x + tau_y*e.y + tau_z*e.z

display(Markdown(r"**Step 1: Translate $\dot{o}$ from linear momentum equation**"))
o_dot = (1/M)*p - me.cross(omega, x_bar_prime)
display(Markdown(r"$\dot{o} = \frac{1}{M}p - (\omega \times \bar{x}')$"))

display(Markdown(r"**Step 2: Derive $\dot{\pi}$ by substituting $\dot{o}$**"))
display(Markdown(r"Starting from Q1 formula: $\dot{\pi} = -M(\dot{o} \times \dot{\bar{x}}) + \tau_e$"))
display(Markdown(r"Since $p = M \dot{\bar{x}}$, we substitute $\dot{\bar{x}} = \frac{p}{M}$"))

# Computationally execute the cross product substitution
# SymPy will automatically cancel the (p/M x p) term since a vector crossed with itself is zero.
pi_dot = -M * me.cross(o_dot, p/M) + tau_e

display(Markdown(r"**$\dot{\pi}$ derived expression:**"))
display(pi_dot)

display(Markdown(r"**Verify against analytical form:** $(\omega \times \bar{x}') \times p + \tau_e$"))
analytical_pi_dot = me.cross(me.cross(omega, x_bar_prime), p) + tau_e
difference = sm.simplify(pi_dot.to_matrix(e) - analytical_pi_dot.to_matrix(e))
display(Markdown(r"*Difference between derivation and analytical formula (Should be 0 matrix):*"))
display(difference)


display(Markdown(r"### Q4.2: Moments about the Center of Mass"))
display(Markdown(r"If the origin is the center of mass, $\bar{x}' = 0$"))
subs_dict = {xbx: 0, xby: 0, xbz: 0}

o_dot_cm = o_dot.subs(subs_dict)
display(Markdown(r"**$\dot{o}$ simplifies to:**"))
display(o_dot_cm)

pi_dot_cm = pi_dot.subs(subs_dict)
display(Markdown(r"**$\dot{\pi}$ simplifies to:**"))
display(pi_dot_cm)


display(Markdown(r"### Q4: The $\mathbf{b}$-frame Description"))

# Define body angular velocity
W1, W2, W3 = me.dynamicsymbols('W1 W2 W3')
Omega = W1*b.x + W2*b.y + W3*b.z
b.set_ang_vel(e, Omega)
Omega_dot = Omega.dt(b)

# Define X_bar (CM relative to O') constant in body frame
X1, X2, X3 = sm.symbols('X1 X2 X3')
X_bar = X1*b.x + X2*b.y + X3*b.z

display(Markdown(r"#### 1. Translational Equation"))
# Let RT_o_ddot represent the R^T * o_ddot term in the b-frame
RT_o_ddot_x, RT_o_ddot_y, RT_o_ddot_z = me.dynamicsymbols('RT_odx RT_ody RT_odz')
RT_o_ddot = RT_o_ddot_x*b.x + RT_o_ddot_y*b.y + RT_o_ddot_z*b.z

display(Markdown(r"**Step 1: Express absolute acceleration in $\mathbf{b}$-frame ($R^T \ddot{x}$)**"))
# Taking the inertial derivative of relative position yields the transport terms
a_cm_body = RT_o_ddot + me.cross(Omega_dot, X_bar) + me.cross(Omega, me.cross(Omega, X_bar))
display(Markdown(r"$R^T \ddot{\bar{x}} =$"))
display(a_cm_body)

display(Markdown(r"**Step 2: Rearrange $M(R^T \ddot{\bar{x}}) = F^e$ to match assignment format**"))
display(Markdown(r"$M(R^T \ddot{o}) + M(\dot{\Omega} \times \bar{X}) = -M(\Omega \times (\Omega \times \bar{X})) + F^e$"))
display(Markdown(r"*Notice that $(\dot{\Omega} \times \bar{X})$ mathematically evaluates exactly to $-(\bar{X} \times \dot{\Omega})$, which matches the $-M \widehat{\bar{X}} \dot{\Omega}$ term.*[cite: 1]"))


display(Markdown(r"#### 2. Rotational Equation"))
# Central inertia tensor
Ic11, Ic22, Ic33 = sm.symbols('Ic11 Ic22 Ic33')
I_c = me.inertia(b, Ic11, Ic22, Ic33)

display(Markdown(r"**Step 1: Local Angular Momentum ($\pi_c$)**"))
pi_c = I_c.dot(Omega)
display(Markdown(r"$\pi_c = \mathbb{I}_c \Omega$"))
display(pi_c)

display(Markdown(r"**Step 2: Time derivative of $\pi_c$ in the inertial frame ($\mathbf{e}$)**"))
display(Markdown(r"SymPy applies the transport theorem: $\dot{\pi}_c = \frac{d}{dt}\pi_c + \Omega \times \pi_c$"))
pi_c_dot = pi_c.dt(e)

display(Markdown(r"**Resulting in Euler's rotational equations (which equals $T_c^e$):**[cite: 1]"))
display(pi_c_dot)
```

# Q#5 - Purely rotating Rigid Bodies

Show that the equations of motion of a purely rotating rigid body with the origin of the body frame $\mathbf{b}$ fixed to the pivot point is given by


\begin{align}
\dot{p}&=M\,R\left(\widehat{\Omega}^2\bar{X} +\dot{\widehat{\Omega}}\bar{X}\right)=  R F^e=f^e,\\
\dot{\pi}&=R\left(\mathbb{I}\dot{\Omega}-\mathbb{I}{\Omega}\times \Omega\right)=\tau^e,
\end{align}
where
\begin{align}
M&\triangleq \sum_{i=1}^nm_i,\\
\mathbb{I}&\triangleq \sum_{i=1}^n\mathbb{I}_i =\sum_{i=1}^n -m_i\widehat{X}_i^2=\sum_{i=1}^n m_i\left(||X_i||^2I_{3\times 3} - X_iX_i^T\right),\\
f^e&=\sum_{i=1}^nf_i^e,\\
\tau^e&=\sum_{i=1}^n(x_i-o)\times f_i^e
\end{align}

```python
import sympy as sm
import sympy.physics.mechanics as me
from IPython.display import display, Markdown

# Enable LaTeX rendering for clean mathematical equations in Jupyter
me.init_vprinting(use_latex=True)
t = me.dynamicsymbols._t

e = me.ReferenceFrame('e') # Inertial frame
b = me.ReferenceFrame('b') # Body frame

M = sm.symbols('M')

display(Markdown(r"### Q5: Purely Rotating Rigid Bodies"))
display(Markdown(r"*Condition: Origin of body frame ($O'$) is fixed to the pivot point.*"))

# Set angular velocity of body frame (Omega)
W1, W2, W3 = me.dynamicsymbols('W1 W2 W3')
Omega = W1*b.x + W2*b.y + W3*b.z
b.set_ang_vel(e, Omega)
Omega_dot = Omega.dt(b)

# Position of Center of Mass (X_bar) is constant in body frame
X1, X2, X3 = sm.symbols('X1 X2 X3')
X_bar = X1*b.x + X2*b.y + X3*b.z

display(Markdown(r"#### 1. Linear Momentum Derivative ($\dot{p}$)"))
display(Markdown(r"**Step 1: Velocity of CM ($v_{cm} = \Omega \times \bar{X}$)**"))
v_cm = me.cross(Omega, X_bar)
display(v_cm)

display(Markdown(r"**Step 2: Acceleration of CM in inertial frame (using Transport Theorem)**"))
display(Markdown(r"$a_{cm} = \frac{d}{dt}v_{cm} + \Omega \times v_{cm}$"))

term1 = me.cross(Omega_dot, X_bar)
display(Markdown(r"*Euler Acceleration ($\dot{\Omega} \times \bar{X}$), representing the $\dot{\widehat{\Omega}}\bar{X}$ term:*[cite: 1]"))
display(term1)

term2 = me.cross(Omega, me.cross(Omega, X_bar))
display(Markdown(r"*Centrifugal Acceleration ($\Omega \times (\Omega \times \bar{X})$), representing the $\widehat{\Omega}^2\bar{X}$ term:*[cite: 1]"))
display(term2)

display(Markdown(r"**Step 3: Total $\dot{p}$ inside the body frame**"))
p_dot_body = M * (term1 + term2)
display(Markdown(r"$\dot{p}_{body} = M [(\dot{\Omega} \times \bar{X}) + (\Omega \times (\Omega \times \bar{X}))]$"))
display(p_dot_body)
display(Markdown(r"*Note: Multiplying this entire term by $R$ maps it to the inertial frame, yielding $\dot{p} = f^e$.*[cite: 1]"))


display(Markdown(r"#### 2. Angular Momentum Derivative ($\dot{\pi}$)"))
# Define Inertia dyadic I
I11, I22, I33 = sm.symbols('I11 I22 I33')
I = me.inertia(b, I11, I22, I33)

display(Markdown(r"**Step 1: Angular Momentum ($\pi = \mathbb{I} \Omega$)**"))
pi = I.dot(Omega)
display(pi)

display(Markdown(r"**Step 2: Differentiate $\pi$ in the inertial frame**"))
display(Markdown(r"Transport theorem: $\dot{\pi} = \frac{d}{dt}\pi + \Omega \times \pi$"))

# d(pi)/dt_b is simply I * Omega_dot since I is constant in the body frame
pi_local_deriv = I.dot(Omega_dot)
display(Markdown(r"*Local derivative ($\mathbb{I} \dot{\Omega}$):*"))
display(pi_local_deriv)

# The gyroscopic term is Omega x (I * Omega)
gyroscopic_term = me.cross(Omega, pi)
display(Markdown(r"*Gyroscopic term ($\Omega \times (\mathbb{I} \Omega)$):*"))
display(gyroscopic_term)

display(Markdown(r"**Step 3: Final $\dot{\pi}$ expression in the body frame**"))
pi_dot_body = pi_local_deriv + gyroscopic_term
display(Markdown(r"$\dot{\pi}_{body} = \mathbb{I} \dot{\Omega} + \Omega \times (\mathbb{I} \Omega)$"))
display(pi_dot_body)

display(Markdown(r"**Mathematical equivalence check:**"))
display(Markdown(r"The assignment formats the gyroscopic term as: $-(\mathbb{I} \Omega) \times \Omega$[cite: 1]"))
display(Markdown(r"In vector algebra, $A \times B = -(B \times A)$."))
display(Markdown(r"Therefore, $\Omega \times (\mathbb{I} \Omega)$ is exactly equivalent to $-(\mathbb{I} \Omega) \times \Omega$."))
display(Markdown(r"*Multiplying the entire body-frame expression by $R$ yields $\dot{\pi} = \tau^e$.*[cite: 1]"))
```
